```python
# текст запроса
query_5_1 = f'''
WITH ranked_employers AS (
  SELECT
    ROW_NUMBER() OVER (ORDER BY COUNT(DISTINCT v.id) DESC) AS position,
    e.name AS employer_name,
    COUNT(DISTINCT v.id) AS cnt
  FROM
    vacancies AS v
  INNER JOIN employers AS e ON v.employer_id = e.id
  GROUP BY e.name
)
SELECT
  position,
  employer_name,
  cnt
FROM ranked_employers
WHERE position IN (1, 5)
ORDER BY position;
'''
```

```python
# текст запроса
query_5_2 = f'''
WITH area_vac AS (
  SELECT
    a.id,
    a.name AS area_name,
    COUNT(v.id) AS vac_cnt
  FROM
    areas AS a
  LEFT JOIN vacancies AS v ON a.id = v.area_id
  GROUP BY a.id, a.name
)
SELECT
  av.id,
  av.area_name,
  av.vac_cnt,
  a.empl_cnt
FROM area_vac AS av
JOIN (
  SELECT
    a.id,
    COUNT(e.id) AS empl_cnt
  FROM areas AS a
  LEFT JOIN employers AS e ON a.id = e.area
  GROUP BY a.id
) AS a ON av.id = a.id
WHERE vac_cnt = 0
ORDER BY empl_cnt DESC
'''
```

```python
import requests
import json
from bs4 import BeautifulSoup

params = {
    'action': 'parse',
    'page': 'Города-миллионеры_России',
    'format': 'json',
    'prop': 'text'
}

api_url = "https://ru.wikipedia.org/w/api.php"

response = requests.get(api_url, params=params,
                       headers={'User-Agent': 'MyBot/1.0'})

if response.status_code == 200:
    data = response.json()
    html_content = data['parse']['text']['*']

    soup = BeautifulSoup(html_content, 'html.parser')
    table = soup.find('table', {'class': 'standard'})

    cities = []
    if table:
        for row in table.find_all('tr')[1:17]:
            columns = row.find_all('td')
            if len(columns) > 1:
                city_name = columns[1].get_text(strip=True)
                cities.append(city_name)

    print("Список городов-миллионников России:")
    for city in cities:
        print(f"- {city}")

    print(f"\nВсего городов: {len(cities)}")
```

```python
# текст запроса
query_6_6 = f'''
  WITH ds_vacancies AS (
    SELECT
        experience,
        salary_from,
        salary_to,
        -- Расчет зарплаты
        CASE
            WHEN salary_from IS NOT NULL AND salary_to IS NOT NULL
                THEN (salary_from + salary_to) / 2.0
            WHEN salary_from IS NOT NULL THEN salary_from
            WHEN salary_to IS NOT NULL THEN salary_to
        END AS calculated_salary
    FROM vacancies
    WHERE
        -- Data Science фильтр
        (LOWER(name) LIKE '%data scientist%'
      OR LOWER(name) LIKE '%data science%'
      OR LOWER(name) LIKE '%исследователь данных%'
      OR (name LIKE '%ML%' AND NOT name LIKE '%HTML%')
      OR LOWER(name) LIKE '%machine learning%'
      OR LOWER(name) LIKE '%машинн%обучен%')
        -- Условие о зарплате
        AND (salary_from IS NOT NULL OR salary_to IS NOT NULL)
)
SELECT
    experience,
    COUNT(*) as vacancy_count,
    ROUND(AVG(calculated_salary)) as avg_salary,
    ROUND(MIN(calculated_salary)) as min_salary,
    ROUND(MAX(calculated_salary)) as max_salary
FROM ds_vacancies
GROUP BY experience
ORDER BY
    CASE experience
        WHEN 'Нет опыта' THEN 1
        WHEN 'От 1 года до 3 лет' THEN 2
        WHEN 'От 3 до 6 лет' THEN 3
        WHEN 'Более 6 лет' THEN 4
        ELSE 5
    END;
'''
```

```python
# результат запроса
df_6_6 = pd.read_sql_query(query_6_6, connection)
df_6_6
```
